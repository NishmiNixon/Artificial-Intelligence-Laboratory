```python

# Experiment 5: 8-Queens Problem Using Backtracking

N = 8

# Store the column position of the queen in each row
board = [-1] * N

# Count the number of placement attempts
attempts = 0


# Check whether a queen can be placed safely
def is_safe(row, col):
    for prev_row in range(row):
        prev_col = board[prev_row]

        # Check same column
        if prev_col == col:
            return False

        # Check diagonals
        if abs(prev_col - col) == abs(prev_row - row):
            return False

    return True


# Solve using recursive backtracking
def solve(row):
    global attempts

    # All 8 queens have been placed
    if row == N:
        return True

    # Try each column in the current row
    for col in range(N):
        attempts += 1

        if is_safe(row, col):
            board[row] = col

            # Place the next queen
            if solve(row + 1):
                return True

            # Backtrack if no solution is found
            board[row] = -1

    return False


# Display the chessboard
def display_board():
    print("\nChessboard Arrangement:\n")

    for row in range(N):
        for col in range(N):
            if board[row] == col:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

    print("\nQueen Positions (Row, Column):")
    for row in range(N):
        print(f"Queen {row + 1}: ({row + 1}, {board[row] + 1})")


# Main program
if solve(0):
    display_board()
    print("\nTotal Placement Attempts:", attempts)
    print("Solution found successfully!")
else:
    print("No solution exists.")
```

```python

# Experiment 6: Map Colouring Problem Using CSP

# 1. Define variables and their domains
regions = ['A', 'B', 'C', 'D', 'E', 'F']

colours = ['Red', 'Green', 'Blue']

# 2. Define the adjacency constraints
neighbours = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['C', 'D', 'F'],
    'F': ['D', 'E']
}

# Store the colour assigned to each region
assignment = {}

# Count colour assignment attempts
attempts = 0


# 3. Check whether a colour is valid
def is_safe(region, colour):
    for neighbour in neighbours[region]:
        if assignment.get(neighbour) == colour:
            return False
    return True


# 4. Solve using recursive backtracking
def solve(index):
    global attempts

    # All regions have been coloured
    if index == len(regions):
        return True

    region = regions[index]

    # Try each colour in the domain
    for colour in colours:
        attempts += 1

        if is_safe(region, colour):
            assignment[region] = colour

            if solve(index + 1):
                return True

            # Backtrack
            del assignment[region]

    return False


# 5. Verify all adjacency constraints
def verify_solution():
    for region in regions:
        for neighbour in neighbours[region]:
            if assignment[region] == assignment[neighbour]:
                return False
    return True


# 6. Execute the CSP solver
print("Map Colouring Problem Using CSP")
print("--------------------------------")

if solve(0):
    print("\nFinal Colour Assignment:")
    for region in regions:
        print(f"Region {region}: {assignment[region]}")

    print("\nAdjacency Constraint Verification:")
    if verify_solution():
        print("All constraints satisfied.")
    else:
        print("Constraint violation detected.")

    print("\nTotal Assignment Attempts:", attempts)
    print("Valid solution found!")
else:
    print("No valid colouring exists.")
    print("Total Assignment Attempts:", attempts)
```

```python

# Experiment 7: Implementing the Minimax Algorithm

# Represent the game tree
game_tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [3, 5],
    'E': [2, 9],
    'F': [12, 5],
    'G': [23, 23]
}

# Store evaluated values of nodes
evaluated_values = {}

# Count terminal nodes evaluated
nodes_evaluated = 0


# Implement Minimax algorithm
def minimax(node, is_maximizing):
    global nodes_evaluated

    # If the node is a terminal utility value
    if isinstance(node, int):
        nodes_evaluated += 1
        return node

    values = []

    # Evaluate all children recursively
    for child in game_tree[node]:
        value = minimax(child, not is_maximizing)
        values.append(value)

    # MAX player selects the highest value
    if is_maximizing:
        best_value = max(values)
    else:
        # MIN player selects the lowest value
        best_value = min(values)

    evaluated_values[node] = best_value
    return best_value


# Execute Minimax from the root MAX node
best_value = minimax('A', True)

# Display evaluated values
print("Minimax Algorithm")
print("-----------------")
print("\nEvaluated Node Values:")

for node in ['D', 'E', 'F', 'G', 'B', 'C', 'A']:
    print(f"Node {node}: {evaluated_values[node]}")

# Select the AI's best move
best_move = max(
    game_tree['A'],
    key=lambda child: evaluated_values[child]
)

print("\nBest Move for AI:", best_move)
print("Optimal Utility Value:", best_value)
print("Terminal Nodes Evaluated:", nodes_evaluated)
```

```python
# Global counters
nodes_evaluated = 0
branches_pruned = 0


def alpha_beta(node, depth, alpha, beta, maximizing_player):
    global nodes_evaluated, branches_pruned

    # Terminal node: return its utility value
    if isinstance(node, int):
        nodes_evaluated += 1
        return node

    if maximizing_player:
        max_eval = float('-inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, depth + 1, alpha, beta, False)
            max_eval = max(max_eval, value)
            alpha = max(alpha, max_eval)

            # Beta cutoff
            if alpha >= beta:
                branches_pruned += len(node) - i - 1
                break

        return max_eval

    else:
        min_eval = float('inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, depth + 1, alpha, beta, True)
            min_eval = min(min_eval, value)
            beta = min(beta, min_eval)

            # Alpha cutoff
            if beta <= alpha:
                branches_pruned += len(node) - i - 1
                break

        return min_eval


# Game tree based on the given diagram
game_tree = [
    [[3, 5], [6, 9], [1, 2]],       # B: MIN
    [[4, 6], [7, 8], [9, 10]],      # C: MIN
    [[1, 3], [12, 14], [15, 18]]    # D: MIN
]

# Root A is a MAX node
alpha = float('-inf')
beta = float('inf')

best_value = float('-inf')
best_move = None

for i, child in enumerate(game_tree):
    value = alpha_beta(child, 1, alpha, beta, False)

    if value > best_value:
        best_value = value
        best_move = chr(ord('B') + i)

    alpha = max(alpha, best_value)

print("ALPHA-BETA PRUNING RESULTS")
print("--------------------------")
print("Selected Move:", best_move)
print("Optimal Utility Value:", best_value)
print("Terminal Nodes Evaluated:", nodes_evaluated)
print("Branches Pruned:", branches_pruned)
```
