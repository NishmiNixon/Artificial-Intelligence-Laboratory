```python
# Forward Chaining - Medical Appointment Eligibility

# Initial Facts
facts = {
    "fever",
    "high_temperature",
    "severe_pain"
}

# Inference Rules
rules = [
    ({"fever", "high_temperature"}, "infection"),
    ({"infection", "severe_pain"}, "priority_consultation"),
    ({"priority_consultation"}, "eligible")
]

print("Initial Facts:")
print(facts)

# Forward Chaining
new_facts = []

while True:
    added = False

    for conditions, conclusion in rules:

        if conditions.issubset(facts) and conclusion not in facts:

            facts.add(conclusion)
            new_facts.append(conclusion)
            print("\nRule Applied:", conditions, "->", conclusion)
            added = True

    if not added:
        break

# Display Results
print("\nNewly Inferred Facts:")
print(new_facts)

print("\nFinal Set of Facts:")
print(facts)

if "eligible" in facts:
    print("\nPatient is eligible for priority consultation.")
else:
    print("\nPatient is not eligible for priority consultation.")
```

```python
# Resolution Refutation - Happy Student

# Knowledge Base in CNF
clauses = [
    {"~E", "~F", "H"},  # E AND F -> H
    {"E"},               # Student enjoys learning
    {"F"},               # Student has free time
    {"~H"}               # Negation of goal
]

print("Initial Knowledge Base:")
for c in clauses:
    print(c)

# Resolution Function
def resolve(c1, c2):
    new_clauses = []

    for literal in c1:
        if literal.startswith("~"):
            opposite = literal[1:]
        else:
            opposite = "~" + literal

        if opposite in c2:
            new_clause = (c1 - {literal}) | (c2 - {opposite})

            if not any(
                lit.startswith("~") and lit[1:] in new_clause
                or "~" + lit in new_clause
                for lit in new_clause
            ):
                new_clauses.append(new_clause)

    return new_clauses


# Resolution Process
print("\nResolution Process:")

while True:
    new = []

    for i in range(len(clauses)):
        for j in range(i + 1, len(clauses)):

            resolvents = resolve(clauses[i], clauses[j])

            for clause in resolvents:
                if clause not in clauses and clause not in new:
                    new.append(clause)
                    print("New Clause:", clause)

    if not new:
        break

    if set() in new:
        print("\nEmpty Clause Obtained!")
        print("Contradiction Established.")
        print("Therefore, the student is HAPPY.")
        break

    clauses.extend(new)

else:
    print("No contradiction found.")
```

```python
# Family Tree
```

```python
# Family Tree using Facts and Rules

# Parent Facts
parent = {
    ("John", "David"),
    ("Mary", "David"),
    ("David", "Alex"),
    ("Sarah", "Alex"),
    ("David", "Emma"),
    ("Sarah", "Emma"),
    ("Alex", "Peter")
}

# Function to find parents
def find_parents(person):
    result = []

    for p, child in parent:
        if child == person:
            result.append(p)

    return result


# Function to find children
def find_children(person):
    result = []

    for p, child in parent:
        if p == person:
            result.append(child)

    return result


# Function to find grandparents
def find_grandparents(person):
    result = []

    for p, child in parent:
        if child == person:
            for gp, parent_name in parent:
                if parent_name == p:
                    result.append(gp)

    return result


# Input
name = input("Enter the person's name: ")

# Display Parents
print("\nParents of", name, ":")
parents = find_parents(name)
print(parents if parents else "No parents found")

# Display Children
print("\nChildren of", name, ":")
children = find_children(name)
print(children if children else "No children found")

# Display Grandparents
print("\nGrandparents of", name, ":")
grandparents = find_grandparents(name)
print(grandparents if grandparents else "No grandparents found")
```

```python
# Pattern Matching - Healthcare Symptom Assessment

# Predefined Symptom Patterns
symptoms = {
    "fever": ["fever", "high temperature", "body heat"],
    "cold": ["cold", "sneezing", "runny nose"],
    "headache": ["headache", "head pain"],
    "cough": ["cough", "dry cough", "sore throat"],
    "stomach_pain": ["stomach pain", "abdominal pain", "stomach ache"]
}

# Healthcare Guidance Rules
guidance = {
    "fever": "Monitor your temperature and consult a doctor if needed.",
    "cold": "Take adequate rest and stay hydrated.",
    "headache": "Take rest and consult a doctor if the pain persists.",
    "cough": "Stay hydrated and seek medical advice if symptoms worsen.",
    "stomach_pain": "Consult a healthcare professional for assessment."
}

# Get Patient Input
user_input = input("Enter your symptoms: ").lower()

# Pattern Matching
matched_symptoms = []

for symptom, keywords in symptoms.items():

    for keyword in keywords:

        if keyword in user_input:
            matched_symptoms.append(symptom)
            break

# Display Results
print("\nSymptom Assessment Result:")

if matched_symptoms:

    for symptom in matched_symptoms:
        print("\nDetected Symptom:", symptom)
        print("Guidance:", guidance[symptom])

else:
    print("No matching symptoms found.")
    print("Please consult a healthcare professional.")

print("\nNote: This is only preliminary guidance, not a medical diagnosis.")
```
