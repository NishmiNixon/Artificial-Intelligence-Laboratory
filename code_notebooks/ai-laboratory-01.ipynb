```python

from collections import deque

# 1. Represent the hospital layout as an adjacency list
hospital = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'E'],
    'D': ['B', 'G'],
    'E': ['B', 'C', 'F'],
    'F': ['E', 'H'],
    'G': ['D', 'H'],
    'H': ['F', 'G']
}

# 2. Implement Breadth-First Search
def bfs(graph, source, destination):
    queue = deque([source])
    visited = {source}
    parent = {source: None}
    nodes_explored = 0

    while queue:
        current = queue.popleft()
        nodes_explored += 1

        # Destination found
        if current == destination:
            break

        # Visit all adjacent nodes
        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append(neighbor)

    # 3. Reconstruct the shortest path
    if destination not in parent:
        print("No route exists between the given locations.")
        return

    path = []
    node = destination

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    print("\nShortest Route:", " -> ".join(path))
    print("Number of Corridors:", len(path) - 1)
    print("Number of Nodes Explored:", nodes_explored)

# 4. Accept source and destination
print("Hospital Navigation System")
print("A - Emergency Entrance")
print("B - Reception")
print("C - Radiology")
print("D - Laboratory")
print("E - Pharmacy")
print("F - ICU")
print("G - Ward")
print("H - Critical-Care Unit")

source = input("\nEnter source node: ").strip().upper()
destination = input("Enter destination node: ").strip().upper()

# 5. Validate input and execute BFS
if source not in hospital or destination not in hospital:
    print("Invalid node! Please enter a node from A to H.")
else:
    bfs(hospital, source, destination)
```

```python

# Experiment 2: DFS for Dependency and Reachability Analysis

# 1. Represent the AI pipeline as a directed adjacency list
pipeline = {
    'A': ['B', 'C', 'D'],  # Data Ingestion
    'B': ['E'],            # Data Preprocessing
    'C': ['F'],            # Feature Engineering
    'D': ['G'],            # Model Training
    'E': ['H'],            # Model Validation
    'F': ['H'],            # Hyperparameter Tuning
    'G': ['H'],            # Model Evaluation
    'H': ['L'],            # Deployment
    'L': []                # Monitoring
}

# 2. Implement Depth-First Search
def dfs(graph, start, target):
    visited = set()
    traversal = []
    parent = {start: None}
    stack = [start]

    while stack:
        current = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        traversal.append(current)

        # Stop when the target is found
        if current == target:
            break

        # Add unvisited neighbors to the stack
        for neighbor in reversed(graph[current]):
            if neighbor not in visited and neighbor not in parent:
                parent[neighbor] = current
                stack.append(neighbor)

    # 3. Display traversal order
    print("\nDFS Traversal Order:", " -> ".join(traversal))

    # 4. Check reachability and reconstruct path
    if target in visited:
        path = []
        node = target

        while node is not None:
            path.append(node)
            node = parent[node]

        path.reverse()

        print("Target is reachable.")
        print("Valid Path:", " -> ".join(path))
    else:
        print("Target is not reachable from the starting task.")


# 5. Accept starting and target tasks
print("AI Deployment Pipeline")
print("A - Data Ingestion")
print("B - Data Preprocessing")
print("C - Feature Engineering")
print("D - Model Training")
print("E - Model Validation")
print("F - Hyperparameter Tuning")
print("G - Model Evaluation")
print("H - Deployment")
print("L - Monitoring")

start = input("\nEnter starting task: ").strip().upper()
target = input("Enter target task: ").strip().upper()

# Validate input
if start not in pipeline or target not in pipeline:
    print("Invalid task! Please enter a valid task label.")
else:
    dfs(pipeline, start, target)
```

```python

from collections import deque

# Capacities of the containers
CAPACITY_A = 4
CAPACITY_B = 3

# Goal: exactly 2 litres in the 4-litre container
GOAL = 2


# Generate all possible states from the current state
def get_next_states(state):
    a, b = state
    next_states = []

    # 1. Fill container A
    if a < CAPACITY_A:
        next_states.append(((CAPACITY_A, b), "Fill 4L container"))

    # 2. Fill container B
    if b < CAPACITY_B:
        next_states.append(((a, CAPACITY_B), "Fill 3L container"))

    # 3. Empty container A
    if a > 0:
        next_states.append(((0, b), "Empty 4L container"))

    # 4. Empty container B
    if b > 0:
        next_states.append(((a, 0), "Empty 3L container"))

    # 5. Pour water from A to B
    if a > 0 and b < CAPACITY_B:
        transfer = min(a, CAPACITY_B - b)
        next_states.append(
            ((a - transfer, b + transfer), "Pour 4L -> 3L")
        )

    # 6. Pour water from B to A
    if b > 0 and a < CAPACITY_A:
        transfer = min(b, CAPACITY_A - a)
        next_states.append(
            ((a + transfer, b - transfer), "Pour 3L -> 4L")
        )

    return next_states


# BFS to find a solution
def water_jug_bfs():
    initial_state = (0, 0)

    # Queue stores (state, path)
    queue = deque([(initial_state, [])])
    visited = {initial_state}

    while queue:
        current_state, path = queue.popleft()
        a, b = current_state

        # Goal test
        if a == GOAL:
            return path + [(current_state, "Goal reached")]

        # Explore the next possible states
        for next_state, action in get_next_states(current_state):
            if next_state not in visited:
                visited.add(next_state)
                new_path = path + [(current_state, action)]
                queue.append((next_state, new_path))

    return None


# Display the solution
solution = water_jug_bfs()

if solution:
    print("Water Jug Problem using BFS")
    print("Initial State: (0, 0)")
    print("State format: (water in 4L, water in 3L)\n")

    for i, (state, action) in enumerate(solution):
        if i == 0:
            print(f"State: {state}")

        if action == "Goal reached":
            print(f"Goal State: {state}")
        else:
            next_state = solution[i + 1][0]
            print(f"Action: {action}")
            print(f"Next State: {next_state}\n")
else:
    print("No solution found.")
```

```python

import heapq

# 1. Represent the environment as a weighted graph
graph = {
    'A': {'B': 4, 'C': 3},
    'B': {'A': 4, 'C': 2, 'D': 5},
    'C': {'A': 3, 'B': 2, 'D': 6, 'E': 4},
    'D': {'B': 5, 'C': 6, 'E': 1, 'F': 2},
    'E': {'C': 4, 'D': 1, 'F': 3},
    'F': {'D': 2, 'E': 3}
}

# 2. Heuristic values from the given diagram
h = {
    'A': 6,
    'B': 4,
    'C': 5,
    'D': 2,
    'E': 3,
    'F': 0
}

# 3. Implement A* Search
def astar(start, goal):
    # Priority queue stores (f(n), node)
    open_list = [(h[start], start)]

    g_cost = {start: 0}
    parent = {start: None}
    expanded = []

    while open_list:
        f_value, current = heapq.heappop(open_list)

        # Ignore outdated queue entries
        if f_value != g_cost[current] + h[current]:
            continue

        # Avoid expanding a node more than once
        if current in expanded:
            continue

        expanded.append(current)

        # Goal test
        if current == goal:
            break

        # Explore neighboring nodes
        for neighbor, cost in graph[current].items():
            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                parent[neighbor] = current

                f = new_g + h[neighbor]
                heapq.heappush(open_list, (f, neighbor))

    # 4. Reconstruct the route
    if goal not in parent:
        print("No route found.")
        return

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    # 5. Display the results
    print("\nA* Search Results")
    print("-----------------")
    print("Selected Route:", " -> ".join(path))
    print("Total Path Cost:", g_cost[goal])
    print("Node Expansion Order:", " -> ".join(expanded))

    print("\nNode Cost Details:")
    print("Node\tg(n)\th(n)\tf(n)")

    for node in expanded:
        print(f"{node}\t{g_cost[node]}\t{h[node]}\t"
              f"{g_cost[node] + h[node]}")


# 6. Accept the starting and goal nodes
print("Intelligent Route Planning")
print("Available nodes: A, B, C, D, E, F")

start = input("Enter starting node: ").strip().upper()
goal = input("Enter goal node: ").strip().upper()

if start not in graph or goal not in graph:
    print("Invalid node! Please enter a node from A to F.")
else:
    astar(start, goal)
```
