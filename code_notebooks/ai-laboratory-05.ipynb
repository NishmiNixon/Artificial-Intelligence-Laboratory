```python
# ID3 Algorithm - Loan Approval Prediction

import math
from collections import Counter

# Step 1: Create Customer Loan Dataset

data = [
    {"Income": "High", "Employment": "Stable", "Credit": "Good", "Loans": "No", "Status": "Approved"},
    {"Income": "High", "Employment": "Stable", "Credit": "Good", "Loans": "Yes", "Status": "Approved"},
    {"Income": "Medium", "Employment": "Stable", "Credit": "Good", "Loans": "No", "Status": "Approved"},
    {"Income": "Low", "Employment": "Stable", "Credit": "Good", "Loans": "No", "Status": "Approved"},
    {"Income": "High", "Employment": "Unstable", "Credit": "Good", "Loans": "No", "Status": "Approved"},
    {"Income": "Medium", "Employment": "Unstable", "Credit": "Good", "Loans": "No", "Status": "Approved"},
    {"Income": "Low", "Employment": "Stable", "Credit": "Poor", "Loans": "No", "Status": "Rejected"},
    {"Income": "Medium", "Employment": "Unstable", "Credit": "Poor", "Loans": "Yes", "Status": "Rejected"},
    {"Income": "Low", "Employment": "Unstable", "Credit": "Poor", "Loans": "No", "Status": "Rejected"},
    {"Income": "High", "Employment": "Unstable", "Credit": "Poor", "Loans": "Yes", "Status": "Rejected"},
    {"Income": "Medium", "Employment": "Stable", "Credit": "Poor", "Loans": "Yes", "Status": "Rejected"},
    {"Income": "Low", "Employment": "Unstable", "Credit": "Good", "Loans": "Yes", "Status": "Rejected"}
]

attributes = ["Income", "Employment", "Credit", "Loans"]

# Step 2: Calculate Entropy

def entropy(data):

    labels = [row["Status"] for row in data]
    total = len(labels)

    counts = Counter(labels)

    result = 0

    for count in counts.values():
        probability = count / total
        result -= probability * math.log2(probability)

    return result


# Step 3: Calculate Information Gain

def information_gain(data, attribute):

    total_entropy = entropy(data)

    values = set(row[attribute] for row in data)

    weighted_entropy = 0

    for value in values:

        subset = [row for row in data
                  if row[attribute] == value]

        weighted_entropy += (
            len(subset) / len(data)
        ) * entropy(subset)

    return total_entropy - weighted_entropy


# Step 4: Construct Decision Tree Recursively

def build_tree(data, attributes):

    labels = [row["Status"] for row in data]

    # If all samples belong to the same class
    if len(set(labels)) == 1:
        return labels[0]

    # If no attributes remain
    if not attributes:
        return Counter(labels).most_common(1)[0][0]

    # Select attribute with highest information gain
    gains = {
        attr: information_gain(data, attr)
        for attr in attributes
    }

    best_attribute = max(gains, key=gains.get)

    tree = {
        "attribute": best_attribute,
        "branches": {},
        "default": Counter(labels).most_common(1)[0][0]
    }

    # Split dataset based on selected attribute
    values = set(row[best_attribute] for row in data)

    remaining_attributes = [
        attr for attr in attributes
        if attr != best_attribute
    ]

    for value in values:

        subset = [
            row for row in data
            if row[best_attribute] == value
        ]

        tree["branches"][value] = build_tree(
            subset, remaining_attributes
        )

    return tree


# Step 5: Display Decision Tree

def display_tree(tree, level=0):

    if isinstance(tree, str):
        print("  " * level + "-> " + tree)
        return

    attribute = tree["attribute"]

    print("  " * level + attribute)

    for value, subtree in tree["branches"].items():

        print("  " * (level + 1) + value)

        display_tree(subtree, level + 2)


# Step 6: Predict New Loan Application

def predict(tree, sample):

    if isinstance(tree, str):
        return tree

    attribute = tree["attribute"]

    value = sample.get(attribute)

    if value not in tree["branches"]:
        return tree["default"]

    return predict(tree["branches"][value], sample)


# Main Program

print("CUSTOMER LOAN APPROVAL USING ID3")

print("\nInitial Entropy:", round(entropy(data), 4))

print("\nInformation Gain:")

for attr in attributes:
    print(attr, ":", round(information_gain(data, attr), 4))

# Build Decision Tree
decision_tree = build_tree(data, attributes)

print("\nGenerated Decision Tree:")
display_tree(decision_tree)

# Test New Application
new_customer = {
    "Income": "High",
    "Employment": "Stable",
    "Credit": "Good",
    "Loans": "No"
}

print("\nNew Customer Details:")
print(new_customer)

result = predict(decision_tree, new_customer)

print("\nLoan Application Result:", result)
```

```python
# Monkey and Banana Problem using State-Space Planning

# Initial State
monkey_position = "Door"
box_position = "Window"
monkey_status = "OnFloor"
has_banana = False

print("Initial State:")
print("Monkey Position:", monkey_position)
print("Box Position:", box_position)
print("Monkey Status:", monkey_status)
print("Has Banana:", has_banana)

# Goal-Based Planning

# Step 1: Monkey moves to the box
print("\n1. Monkey moves towards the box.")
monkey_position = "Window"

# Step 2: Monkey pushes the box under the banana
print("2. Monkey pushes the box under the banana.")
box_position = "Banana"

# Step 3: Monkey climbs onto the box
print("3. Monkey climbs onto the box.")
monkey_status = "OnBox"

# Step 4: Monkey grasps the banana
if monkey_position == box_position and monkey_status == "OnBox":
    print("4. Monkey grasps the banana.")
    has_banana = True

# Final State
print("\nFinal State:")
print("Monkey Position:", monkey_position)
print("Box Position:", box_position)
print("Monkey Status:", monkey_status)
print("Has Banana:", has_banana)

# Check Goal State
if has_banana:
    print("\nGoal Achieved! Monkey has obtained the banana.")
else:
    print("\nGoal Not Achieved!")
```

```python
# Towers of Hanoi using Recursion

def tower_of_hanoi(n, source, auxiliary, destination):

    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return

    # Step 1: Move n-1 disks to Auxiliary rod
    tower_of_hanoi(n - 1, source, destination, auxiliary)

    # Step 2: Move largest disk to Destination rod
    print("Move disk", n, "from", source, "to", destination)

    # Step 3: Move n-1 disks to Destination rod
    tower_of_hanoi(n - 1, auxiliary, source, destination)


# Main Program
n = int(input("Enter the number of disks: "))

print("\nSteps to solve Towers of Hanoi:\n")

tower_of_hanoi(n, 'A', 'B', 'C')

print("\nTotal number of moves:", 2**n - 1)
```

```python
# Simple Reflex Agent - Vacuum World

# Initial Room Conditions
rooms = {
    'A': 'Dirty',
    'B': 'Dirty',
    'C': 'Clean',
    'D': 'Dirty'
}

# Initial Agent Location
location = 'A'

print("Initial Room Conditions:")
print(rooms)

print("\nVacuum Cleaning Simulation:\n")

# Agent Simulation
while True:

    # Step 1: Identify Current Location
    print("Current Location:", location)

    # Step 2: Check Room Condition
    if rooms[location] == 'Dirty':

        # Step 3: Clean the Room
        print("Room is Dirty -> Action: SUCK")
        rooms[location] = 'Clean'

    else:

        print("Room is Clean -> Action: MOVE")

        # Step 4: Move to Next Room
        if location == 'A':
            location = 'B'

        elif location == 'B':
            location = 'C'

        elif location == 'C':
            location = 'D'

        elif location == 'D':
            print("\nAll rooms have been checked!")
            break

    print("Room Status:", rooms)
    print()

    # Stop when all rooms are clean
    if all(status == 'Clean' for status in rooms.values()):
        print("All rooms are clean!")
        break

# Final Result
print("\nFinal Room Conditions:")
print(rooms)

print("\nCleaning Simulation Completed!")
```
